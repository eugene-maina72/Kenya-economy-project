## Data Preparation

From loading the data we can notice a few decrepencies that we need to correct:
* The column names  on all the dataframes (except `gdp_data` and `gni_data`) are incorrect.
* All of the dataframes have data that is redundant. All of the data sets contain years outside our sample size.The scope of the project is to analyse the trends from  the year 2000-2023
* Because of the mismatch in column names, the coumn named `% Annual change` must be dropped on majority of the data sets except the `gdp_data` and `gni_data` 
* These are far too many data sets to deal with as far as ease of use, thus we must merge them depending on the problem question objective. 
* To make this merge possible and make thata more readable we need to extract the year from the `date` columns and use that as the unique key since all the data sets contain the `date` column.Thus we have to `set_index`
* For plotting purposes,we need to convert the GDP columns to full figures meaning we have to multiply the amount by `10**9`
* There are some missing values in various data sets

***

### Data Cleaning

* Lets start by renaming the columns to more appropriate column names.

```python
# Changing the column names to a more appropriate name.
#Renaming population columns
pop_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'Population', ' Per Capita (US $)' : '% Population Growth'}, inplace=True)

#Renaming foreign direct investment columns
fdi_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'FDI Amount(USD)', ' Per Capita (US $)' : "% Change in FDI"}, inplace=True)

#Renaming inflation columns
inflation_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : '% Inflation Rate', ' Per Capita (US $)' : '% Change in Inflation'}, inplace=True)

#Renaming manufacturing output columns
manufacturing_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'Manufacturing Output(USD)', ' Per Capita (US $)' : '% Change in Output'}, inplace=True)

#Renaming trade balance deficit columns
trade_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'Trade Deficit(USD)', ' Per Capita (US $)' : '% Change Deficit'}, inplace=True)

#Renaming gross domestic product columns
gdp_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'GDP Amount(USD)', ' Annual % Change' : '% Change in GDP', ' Per Capita (US $)' : 'GDP per Capita'}, inplace=True)

#Renaming gross national income columns
gni_data.rename(columns={'Date': 'Year', ' GDP (Billions of US $)' : 'GNI Amount(USD)', ' Annual % Change' :'% Change in GNI', ' Per Capita (US $)' : 'GNI per Capita'}, inplace=True)
```

* Now to drop the redundant `% Annual Change` column.

```python
# Removing the redundant column

#Removing from the population data
pop_data.drop(columns=[' Annual % Change'], inplace=True)

#Removing from the foreign direct investment data
fdi_data.drop(columns=[' Annual % Change'], inplace=True)

#Removing from the inflation data
inflation_data.drop(columns=[' Annual % Change'], inplace=True)

#Removing from the manufacturing output data.
manufacturing_data.drop(columns=[' Annual % Change'], inplace=True)

#Removing from the trade balance deficit data.
trade_data.drop(columns=[' Annual % Change'], inplace=True)
```

```python
# Previewing the column changes.
print(pop_data.columns)
print(fdi_data.columns)
print(inflation_data.columns)
print(manufacturing_data.columns)
print(trade_data.columns)
print(gdp_data.columns)
print(gni_data.columns)
```

* Now to change the year column to only display the year (ie. 2000,2001,2002).

```python
# Changing the year column to display the appropriate data type.

#Changing the year column to datetime in the population data.
pop_data['Year'] = pd.to_datetime(pop_data['Year']).dt.year

#Changing the year column to datetime in the foreign direct investment data.
fdi_data['Year'] = pd.to_datetime(fdi_data['Year']).dt.year

#Changing the year column to datetime in the inflation data.
inflation_data['Year'] = pd.to_datetime(inflation_data['Year']).dt.year

#Changing the year column to datetime in the manufacturing output data.
manufacturing_data['Year'] = pd.to_datetime(manufacturing_data['Year']).dt.year

#Changing the year column to datetime in the trade balance deficit data.
trade_data['Year'] = pd.to_datetime(trade_data['Year']).dt.year

#Changing the year column to datetime in the gross domestic product data.
gdp_data['Year'] = pd.to_datetime(gdp_data['Year']).dt.year

#Changing the year column to datetime in the gross national income data.
gni_data['Year'] = pd.to_datetime(gni_data['Year']).dt.year



```

* Narrowing our sample size (2000-2023) and setting index to `"Year"`

```python
# Selecting only data from 2000 to 2023 and setting the year as index.

#Selecting data from 2000 to 2023 in the population data.
pop_data = pop_data[(pop_data['Year'] >= 2000) & (pop_data['Year'] <= 2023)]
pop_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the inflation data.
inflation_data = inflation_data[(inflation_data['Year'] >= 2000) & (inflation_data['Year'] <= 2023)]
inflation_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the population data.
gdp_data = gdp_data[(gdp_data['Year'] >= 2000) & (gdp_data['Year'] <= 2023)]
gdp_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the population data.
gni_data = gni_data[(gni_data['Year'] >= 2000) & (gni_data['Year'] <= 2023)]
gni_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the population data.
fdi_data = fdi_data[(fdi_data['Year'] >= 2000) & (fdi_data['Year'] <= 2023)]
fdi_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the population data.
trade_data = trade_data[(trade_data['Year'] >= 2000) & (trade_data['Year'] <= 2023)]
trade_data.set_index('Year', inplace=True)

#Selecting data from 2000 to 2023 in the population data.
manufacturing_data = manufacturing_data[(manufacturing_data['Year'] >= 2000) & (manufacturing_data['Year'] <= 2023)]
manufacturing_data.set_index('Year', inplace=True)
```

* Now to multiply the values by 1 billion in the `Amount` column.For this we can create a function then use `.apply` to change all values in the column.

```python
def change_to_billion(value):
    """Converting the value to billions"""
    return value * 10**9

# Changing the GDP data to billions
gdp_data['GDP Amount(USD)'] = gdp_data['GDP Amount(USD)'].apply(change_to_billion)
# Changing the GNI data to billions
gni_data['GNI Amount(USD)'] = gni_data['GNI Amount(USD)'].apply(change_to_billion)
# Changing the FDI data to billions
fdi_data['FDI Amount(USD)'] = fdi_data['FDI Amount(USD)'].apply(change_to_billion)
# Changing the trade data to billions
trade_data['Trade Deficit(USD)'] = trade_data['Trade Deficit(USD)'].apply(change_to_billion)
# Changing the manufacturing data to billions
manufacturing_data['Manufacturing Output(USD)'] = manufacturing_data['Manufacturing Output(USD)'].apply(change_to_billion)
```

### Merging the data into a single data frame
* Now we can view the changes that have been done and display them in a combined dataframe sample.After this we can then check for missing values and fill them accordingly.

```python
# We can use the concat method to combine the dataframes.
df_combined = pd.concat([pop_data, trade_data, manufacturing_data, fdi_data, gdp_data, gni_data, inflation_data], axis=1)
df_combined.info()
```

### Dealing with missing values

* The `manufacturing_data` is missing a whole row entry which is all value based so we can use the `.describe` method to determine whether we can use mean or median or mode to fill in the values

```python
df_combined.describe()
```

* To adress the missing row in the `Manufucturing Output` column, I shall fill it with the median due to the presence of large outliers and a large standard deviation.



```python
# Fill the missing value in the data with the median value of the column and check for others.
# Manufacturing Output(USD) median
manufacturing_output_median = manufacturing_data['Manufacturing Output(USD)'].median()
new_manufacturing_row = {'Manufacturing Output(USD)': manufacturing_output_median, '% Change in Output': 0 }

# Get the next index in the dataframe
next_index = manufacturing_data.index[-1] + 1

# Append the new row to the DataFrame
manufacturing_data.loc[next_index] = new_manufacturing_row
```

* For the `% Change` in GNI data and the Manufucturing output data, since it is determined by the percentage difference beteween 2 consecutive values in `GNI Amount(USD)` and `Manufacturing Output` respectively, the best method would to fill the values using the Numpy method `np.where` and `pct_change` to fill the missing values with the percentage differences.

```python
# Calculate the difference between consecutive 'Manufacturing Output' entries using pct_ and return the percentage change.
manufacturing_data['% Change in Output'] = np.where(manufacturing_data['% Change in Output'] == 0, manufacturing_data['Manufacturing Output(USD)'].pct_change(), manufacturing_data['% Change in Output'])

# Calculate the percentage differnce between GNI entries using pct_change and return the percentage change.
gni_data['% Change in GNI'] = np.where(gni_data['% Change in GNI'].isnull(), gni_data['GNI Amount(USD)'].pct_change(), gni_data['% Change in GNI'])
```

```python
# Display info to view changes.
print(manufacturing_data.info())
print(gni_data.info())
```

### Plotting GDP against Population

```python
# Plotting the data side by side to visualize the trends.
# Selecting variables to plot
x = df_combined.index
population = df_combined['Population']
gdp = df_combined['GDP Amount(USD)']

#Plotting table and population data
sns.set_style('whitegrid')
fig, ax1 = plt.subplots(figsize=(10, 6))
plt.title('Population vs GDP', fontsize=15)
ax1.set_xlabel('Year')
ax1.set_ylabel('Population', color='tab:red', fontsize=12)
ax1.plot(x, population, color='tab:red')
ax1.set_xticks(x)

#Plotting GDP data
ax2 = ax1.twinx()
ax2.set_ylabel('GDP Amount(USD)', color='tab:blue', fontsize=12)
ax2.plot(x, gdp, color='tab:blue')

# Display and save the plot
fig.tight_layout()
plt.show()
fig.savefig('images\population_vs_gdp.png')
```

### Plotting Foreign Direct investment vs GDP

```python
# Plotting the two graphs on the same axes

sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(10, 6))
plt.title('Foreign Direct Investment vs GDP', fontsize=15)
ax.set_xlabel('Year')
ax.set_ylabel('Foreign Direct Investment(USD)', color='tab:red', fontsize=12)
ax.plot(x, df_combined['FDI Amount(USD)'], color='tab:red')
ax.set_xticks(df_combined.index)
ax2 = ax.twinx()
ax2.set_ylabel('GDP Amount(USD)', color='tab:blue', fontsize=12)
ax2.plot(x, df_combined['GDP Amount(USD)'], color='tab:blue')
ax.legend(['FDI Amount(USD)', 'GDP Amount(USD)'])
fig.tight_layout()
plt.show()
fig.savefig('images\gdp_vs_fdi.png')

```

### Plotting GNI per Capita vs Inflation rate

```python
# Plotting the two graphs on the same axes(Bar graph and Linegraph)

fig, ax = plt.subplots(figsize=(10, 6))
plt.title('Gross National Index per Capita vs Inflation Rate', fontsize=15)
ax.set_xlabel('Year')
ax.set_ylabel('GNI per Capita', color='tab:green', fontsize=12)
ax.bar(x, df_combined['GNI per Capita'], color='tab:green')
ax.set_xticks(df_combined.index)
ax2 = ax.twinx()
ax2.set_ylabel('Inflation Rate', color='black', fontsize=12)
ax2.plot(x, df_combined['% Inflation Rate'], color='black')
ax.legend(['GNI per Capita', 'Inflation Rate'])
fig.tight_layout()
plt.show()
fig.savefig('images\gni_vs_inflation.png')
```

### Determining Pearson Correlation of Manufucturing output vs Trade Balance Deficit 

```python
# Determining correlation between Manufucturing output and Trade Balance Deficit.

correlation = df_combined['Manufacturing Output(USD)'].corr(df_combined['Trade Deficit(USD)'])

# Defining a function to return a comment based on the correlation value.
def correlation_comment(correlation):
    """Determines the correlation between two variables"""
    if correlation > 0.5:
        return 'Strong Positive Correlation'
    elif correlation < -0.5:
        return 'Strong Negative Correlation'
    else:
        return 'Little to no correlation'
    
# Display the correlation and the comment.
print(f'The correlation between Manafucturing Output and Trade Balance Deficit is {round(correlation, 2)}. \nThis is a {correlation_comment(correlation)}')
```

### Plotting a Graph of Manufacturing Output vs Trade Deficit.

```python
# Defining the variables to be plot
x = df_combined.index
trade_deficit = df_combined['Trade Deficit(USD)']
manufacturing_output = df_combined['Manufacturing Output(USD)']

# Plotting the graph
sns.set_style('whitegrid')
fig, ax = plt.subplots(figsize=(10, 6))
plt.title('Trade Deficit vs Manufacturing Output', fontsize=15)
ax.set_xlabel('Year')

# Plotting the two graphs on the same axes
line1 = plt.plot(x, trade_deficit, color='red', linestyle='-', marker='o', label="Connected Points")
line2 = plt.plot(x, manufacturing_output, color='blue', linestyle='-', marker='o', label="Connected Points")
ax.set_ylabel('Trade Deficit(USD)', color='tab:red', fontsize=12)
ax.set_xticks(df_combined.index)
ax2 = ax.twinx()
ax2.set_ylabel('Manufacturing Output(USD)', color='tab:blue', fontsize=12)
ax.legend(['Trade Deficit(USD)', 'Manufacturing Output(USD)'])

# Display and save the plot
fig.tight_layout()
plt.show()
fig.savefig('images\Trade_deficit_vs_manufacturing_output.png')
```
